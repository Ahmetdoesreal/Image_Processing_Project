# iremimg — ayrıntılı inceleme raporu

Altı dosyanın bütün, sol ve sağ görünümlerini sekiz JPEG kalitesinde inceledik: **5, 10, 20, 40, 60, 80, 90, 95**. Toplam 144 analiz var. Duyuru, karga ve error için de kontrol/render Q20 kopyaları eklendi.

[Etkileşimli ana raporu aç](index.html) · [Kalite 20 dosyaları](Q20/index.html) · [Bütün ölçümler](all_measurements.csv)

Notebook açıklama ve kayıtlı görsellerden oluşur; kod çalıştırmak gerekmez.


## Ne ölçüyoruz?

Girdiyi tekrar JPEG olarak kaydedip önceki ve sonraki piksel değerlerinin mutlak farklarının üç kanal ortalamasına bakıyoruz. Bütün hesaplar, görünümün kendi piksel boyutunda yapıldı.

Kaynakların ikisi AVIF, ikisi WebP ve ikisi JPEG. JPEG dışındaki girdilerde bu, JPEG'e dönüşüm farkıdır. Sol ve sağ taraflar doğrulanmış orijinal/sahte etiketleri değildir. Düzenleme maskeleri verilmediği için tespit doğruluğu hesaplamıyoruz.

Paneller ayrı ayrı JPEG'e çevrildi; bütün kolajdan alınan harita kesitleri değiller. Kenarlar, dokular, ayırıcı çizgiler ve bu ayrı kodlama sonuçları etkileyebilir. Renk ölçeği 0–20; ölçümler ham farklardan alınır.

## Ham verilerle devam etmek için

Her görselin `whole`, `left` ve `right` klasörlerinde `raw_ela_maps.npz` ve `metrics.csv` var. NPZ, q05–q95 haritalarını float32 olarak, renk uygulanmadan saklıyor. Renk üst sınırının etkisini incelemek veya başka bir görünüm üretmek için bunlar kullanılabilir.

Histogram, Canny kenarlar ve Q20 RGB fark ×10 dosyaları ek görsel bağlam sağlar. Kaynağın bir bölümünün daha fazla fark üretmesi tek başına orada düzenleme olduğunu kanıtlamaz.
