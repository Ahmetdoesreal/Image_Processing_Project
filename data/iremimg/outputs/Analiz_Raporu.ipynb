# İrem image analysis report

Six input images were analyzed as whole images and as left and right panels. The report embeds quality-20 figures, with additional qualities and downloads available in the [HTML report](index.html).

## Measurements

For each source and view, the same input pixels were encoded at eight JPEG qualities from 5 to 95. ELA is the mean absolute difference across the color channels. The panel positions describe where crops came from. Conversion and JPEG block alignment can affect the comparisons.
